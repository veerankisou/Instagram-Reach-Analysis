```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.linear_model import LinearRegression, LogisticRegression

data = {
    "likes" : np.random.randint(100,500,1000),
    "comments" : np.random.randint(10,100,1000),
    "shares" : np.random.randint(10,100, 1000),
    "hashtags" : np.random.randint(10,500,1000),
    "reach" : np.random.randint(500, 5000, 1000)
}

df = pd.DataFrame(data)
print(df.head())
print()
print(df.info())
print()
print(df.describe())

sns.pairplot(df)
plt.show()

x = df[["likes"]]
y = df["reach"]

model = LinearRegression()
model.fit(x,y)

plt.scatter(x,y, color = "blue")
plt.plot(x, model.predict(x), color = "red")
plt.xlabel("likes")
plt.ylabel("reach")
plt.title("Likes vs Reach")
plt.show()

df["popular"] = df["reach"] > 2000
print(df)


```

```python
x = df[["likes", "comments", "shares", "hashtags"]]
y = df["reach"]

log_model = LogisticRegression()
log_model.fit(x,y)

df["predicted_popular"] = log_model.predict(x)

print(df)


likes = int(input("Enter the number of likes: "))
comments = int(input("Enter the number of comments: "))
shares = int(input("Enter the number of shares: "))
hashtags = int(input("Enter the number of hashtags: "))


predicted_reach = model.predict([[likes]])[0]
predicted_popularity = log_model.predict([[likes, shares, comments, hashtags]])[0]

print("Predicted Reach: ", predicted_reach)
print("Predicted Popularity", 'Yes' if predicted_popularity else "No")
```
